# VISUALIZE FINAL GRADES 

```python
plt.figure(figsize=(8,5))

plt.hist(df["G3"], bins=10)

plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Mathematics Grades")

plt.show()
```

# Analyze gender distribution

```python
df["sex"].value_counts()                          # COUNT OF MALE & FEMALE 
```

```python
ax = sns.countplot(data=df, x="sex")

plt.title("Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

# Add values on top of bars
ax.bar_label(ax.containers[0])

plt.show()
```

# GENDER VS FINAL GRADE 

```python
df.groupby("sex")["G3"].mean()                 # AVG OF GRADE BY INDIVDUAL SEX 
```

```python
ax = sns.barplot(data=df, x="sex", y="G3")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

ax.bar_label(ax.containers[0], fmt="%.2f")

plt.show()
```

# STUDY TIME VS FINAL GRADE 

```python
df.groupby("studytime")["G3"].mean()                      # AVG FINAL GRADE BY STUDYTIME 
```

```python
sns.barplot(data=df, x="studytime", y="G3")

plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade")

plt.show()
```

# ABSENCE VS FINAL GRADE 

```python
sns.scatterplot(data=df, x="absences", y="G3")

plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")

plt.show()
```

# PREVIOUS GRADE VS FINAL GRADES 

```python
df[["G1", "G2", "G3"]].corr()                           # CALCULATES CORRELATION BTW COLUMNS 
```

```python
sns.heatmap(
    df[["G1", "G2", "G3"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Student Grades")

plt.show()
```

# EXTRA EDUCATIONAL SUPPORT 

```python
df.groupby("schoolsup")["G3"].mean()                     #  AVG OF GRADE BY SCHOOL SUPPORT
```

```python
df.groupby("famsup")["G3"].mean()                        #  AVG OF GRADE BY FAMILY SUPPORT
```

```python
df.groupby("paid")["G3"].mean()                          #  AVG OF GRADE BY PAID CLASSES SUPPORT
```

```python
# CALCULATE AVERAGE G3 FOR EACH SUPPORT CATEGORY 
schoolsup_avg = df.groupby("schoolsup")["G3"].mean()
famsup_avg = df.groupby("famsup")["G3"].mean()
paid_avg = df.groupby("paid")["G3"].mean()


# CREATE A DATAFRAME FOR COMPARISON 
comparison = pd.DataFrame({
    "School Support": schoolsup_avg,
    "Family Support": famsup_avg,
    "Paid Classes": paid_avg
})


ax = comparison.plot(kind="bar", figsize=(10, 6))

plt.title("Comparison of Support and Paid Classes vs Final Grade")
plt.xlabel("Response (No / Yes)")
plt.ylabel("Average Final Grade")
plt.xticks(rotation=0)
plt.legend(title="Support Type")

plt.show()
```
